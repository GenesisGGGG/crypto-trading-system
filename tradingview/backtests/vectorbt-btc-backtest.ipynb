```python
# ───────────────────────────────────────────────────────────
# 1) Import all required libraries
import vectorbt as vbt
import pandas as pd

# ───────────────────────────────────────────────────────────
# 2) Download Bitcoin daily close prices from Yahoo Finance
#    Use the class method YFData.download() which works in your version
btc_price = vbt.YFData.download(
    'BTC-USD', 
    start='2022-01-01', 
    end='2023-01-01'
).get('Close')

# ───────────────────────────────────────────────────────────
# 3) Generate entry and exit signals
#    • entries: buy when price makes a forward breakout over the last 3 days
#    • exits:  sell when price makes a forward swing top over the last 3 days
entries = btc_price.vbt.signals.fbt(n=3)
exits   = btc_price.vbt.signals.fst(n=3)

# ───────────────────────────────────────────────────────────
# 4) Run the backtest
portfolio = vbt.Portfolio.from_signals(
    btc_price, 
    entries, 
    exits
)

# ───────────────────────────────────────────────────────────
# 5) Show performance statistics
print(portfolio.stats())

# ───────────────────────────────────────────────────────────
# 6) Plot equity curve with buy/sell markers
portfolio.plot().show()
# ───────────────────────────────────────────────────────────
```
